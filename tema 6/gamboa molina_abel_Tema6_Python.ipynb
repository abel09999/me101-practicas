```python
import numpy as np
import matplotlib.pyplot as plt

# Completa: construye el degradado de 5x5
gris = np.tile(np.linspace(0, 255, 5), (5, 1))

# Completa: imprime shape, dtype y mean()
print("shape:", gris.shape)
print("dtype:", gris.dtype)
print("mean():", gris.mean())
```

```python
# Completa: visualiza y guarda la imagen
plt.imshow(gris, cmap="gray", vmin=0, vmax=255)
plt.savefig("gris_tema6.png")
plt.show()
```

```python
# Completa: recorte de la mitad derecha
recorte = gris[:, -3:]
print(recorte)
```

```python
# Completa: construye la imagen RGB de dos mitades
rgb = np.zeros((4, 4, 3), dtype=np.uint8)
rgb[:, 0:2] = [255, 0, 0]    # mitad izquierda: rojo (R,G,B)
rgb[:, 2:4] = [0, 0, 255]    # mitad derecha: azul

print(rgb.shape)
print("Pixel [0,0]:", rgb[0, 0])
print("Pixel [0,3]:", rgb[0, 3])
```

```python
# Completa: extrae cada canal
R = rgb[:, :, 0]
G = rgb[:, :, 1]
B = rgb[:, :, 2]

print("Canal R:\n", R)
print("Canal G:\n", G)
print("Canal B:\n", B)
```

```python
# Completa: gris_promedio y gris_luminosidad
# Se convierte a float para evitar el desbordamiento de uint8 al sumar canales
R, G, B = [rgb[:, :, i].astype(float) for i in range(3)]

gris_promedio = (R + G + B) / 3
gris_luminosidad = 0.299 * R + 0.587 * G + 0.114 * B

print("Promedio simple:\n", gris_promedio)
print("Luminosidad ponderada:\n", gris_luminosidad)
```

```python
# Completa: binarizacion con tu propio umbral
umbral = 50
binaria = (gris_luminosidad > umbral).astype(np.uint8)

print(binaria)
```
